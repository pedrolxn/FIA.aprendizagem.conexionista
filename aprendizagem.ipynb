```python
import numpy as np
from numpy.random import randn

# Definição das dimensões: Batch(N), Input(D_in), Hidden(H), Output(D_out)
N, D_in, H, D_out = 64, 1000, 100, 10

# Inicialização aleatória dos dados de entrada (x), saída esperada (y) e pesos (w1, w2)
x, y = randn(N, D_in), randn(N, D_out)
w1, w2 = randn(D_in, H), randn(H, D_out)

for t in range(2000):
    # Forward pass (cálculo das previsões)
    h = 1 / (1 + np.exp(-x.dot(w1)))
    y_pred = h.dot(w2)

    # Cálculo e impressão da função de perda (Loss)
    loss = np.square(y_pred - y).sum()
    print(f"Época {t}: {loss:.4f}")

    # Backpropagation (cálculo dos gradientes)
    grad_y_pred = 2.0 * (y_pred - y)
    grad_w2 = h.T.dot(grad_y_pred)
    grad_h = grad_y_pred.dot(w2.T)
    grad_w1 = x.T.dot(grad_h * h * (1 - h))

    # Atualização dos pesos (Gradient Descent)
    w1 -= 1e-4 * grad_w1
    w2 -= 1e-4 * grad_w2
```
